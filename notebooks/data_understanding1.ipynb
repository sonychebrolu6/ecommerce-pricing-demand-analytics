```python
print("Number of rows:", retail.shape[0])
print("Number of columns:", retail.shape[1])

print("Unique products:", retail["product_id"].nunique())
print("Unique categories:", retail["product_category_name"].nunique())
print("Unique periods:", retail["month_year"].nunique())
```

```python
product_counts = retail.groupby("product_id").size()
print(product_counts.describe())
```

```python
price_variation = retail.groupby("product_id")["unit_price"].nunique()

print("Products with more than one observed price:",
      (price_variation > 1).sum())

print("\nPrice variation statistics:")
print(price_variation.describe())
```

```python
quantity_variation = retail.groupby("product_id")["qty"].nunique()

print("Products with more than one observed quantity:",
      (quantity_variation > 1).sum())

print("\nQuantity variation statistics:")
print(quantity_variation.describe())
```

```python
retail["month_year"] = pd.to_datetime(
    retail["month_year"],
    format="%d-%m-%Y"
)
```

```python
print("Start date:", retail["month_year"].min())
print("End date:", retail["month_year"].max())
```

```python
product_months = retail.groupby("product_id")["month_year"].nunique()
print(product_months.describe())
```

```python
print(product_months.value_counts().sort_index())
```

```python
duplicate_product_month = retail.duplicated(
    subset=["product_id", "month_year"]
).sum()

print("Duplicate product-month records:", duplicate_product_month)
```

```python
print("Quantity <= 0:", (retail["qty"] <= 0).sum())
print("Unit price <= 0:", (retail["unit_price"] <= 0).sum())
print("Total price <= 0:", (retail["total_price"] <= 0).sum())
print("Customers <= 0:", (retail["customers"] <= 0).sum())
```

```python
print("Duplicate rows:", retail.duplicated().sum())
```

```python
product_summary = retail.groupby("product_id").agg(
    observations=("month_year", "count"),
    min_price=("unit_price", "min"),
    max_price=("unit_price", "max"),
    avg_price=("unit_price", "mean"),
    total_quantity=("qty", "sum"),
    avg_quantity=("qty", "mean")
)

product_summary.head(10)
```
